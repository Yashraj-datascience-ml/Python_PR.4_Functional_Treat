```python
#Project 4

print("Welcome to the Student Data Organizer!\n")

students = []     
all_subjects = set()  

while True:
    print("1. Add Student")
    print("2. Display All Students")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display Subjects Offered")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    #  ADD STUDENT 
    if choice == "1":
        print("\n--- Add Student ---")

        name = input("Name: ")
        age = int(input("Age: "))  
        grade = input("Grade: ")

        sub = input("Subjects (comma-separated): ")
        subjects = [s.strip() for s in sub.split(",")]

        # add unique subjects to SET
        for s in subjects:
            all_subjects.add(s)

        sid = input("Student ID: ")
        dob = input("Date of Birth (YYYY-MM-DD): ")

        # tuple for unchangeable info
        unique_info = (sid, dob)

        # dictionary for student
        student = {
            "id_dob": unique_info,
            "name": name,
            "age": age,
            "grade": grade,
            "subjects": subjects
        }

        students.append(student)
        print("Student added!\n")

    #  DISPLAY ALL 
    elif choice == "2":
        print("\n--- All Students ---")
        if len(students) == 0:
            print("No students yet.\n")
        else:
            for s in students:
                sid, dob = s["id_dob"]
                print(f"ID: {sid} | Name: {s['name']} | Age: {s['age']} | Grade: {s['grade']}")
                print("   DOB:", dob)
                print("   Subjects:", ", ".join(s["subjects"]))
                print()
    
    # UPDATE STUDENT 
    elif choice == "3":
        uid = input("\nEnter Student ID to update: ")
        found = False

        for s in students:
            if s["id_dob"][0] == uid:
                found = True
                print("Leave blank if no change.")

                new_age = input("New Age: ")
                if new_age != "":
                    s["age"] = int(new_age)

                new_sub = input("New Subjects: ")
                if new_sub != "":
                    new_list = [x.strip() for x in new_sub.split(",")]
                    s["subjects"] = new_list
                    for x in new_list:
                        all_subjects.add(x)

                print("Updated!\n")
                break
        
        if not found:
            print("Student not found.\n")

    # DELETE STUDENT 
    elif choice == "4":
        uid = input("\nEnter Student ID to delete: ")
        deleted = False

        for i in range(len(students)):
            if students[i]["id_dob"][0] == uid:
                del students[i]   
                print("Deleted!\n")
                deleted = True
                break
        
        if not deleted:
            print("Student not found.\n")

    #  DISPLAY SUBJECTS 
    elif choice == "5":
        print("\n--- Subjects Offered ---")
        if len(all_subjects) == 0:
            print("No subjects yet.\n")
        else:
            for sub in all_subjects:
                print(sub)
            print()

    # EXIT
    elif choice == "6":
        print("Thanks for using Student Data Organizer!")
        break

    else:
        print("Invalid choice!\n")
```
